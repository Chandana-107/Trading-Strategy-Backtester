# Trading Strategy Backtester Example

This Jupyter notebook demonstrates how to use the Trading Strategy Backtester built in Python. We will walk through downloading historical stock data, applying trading strategies, and analyzing the results.

```python
# Import necessary libraries
import yfinance as yf
import pandas as pd
from src.backtester.data.downloader import download_data
from src.backtester.strategies.sma_crossover import SMA_Crossover
from src.backtester.strategies.rsi_strategy import RSI_Strategy
from src.backtester.engine import BacktestEngine
from src.backtester.reporting import generate_report

# Set parameters for the backtest
ticker = 'AAPL'
start_date = '2020-01-01'
end_date = '2023-01-01'

# Download historical data
data = download_data(ticker, start_date, end_date)

# Initialize strategies
sma_strategy = SMA_Crossover(data)
rsi_strategy = RSI_Strategy(data)

# Run backtests
sma_results = BacktestEngine(sma_strategy).run()
rsi_results = BacktestEngine(rsi_strategy).run()

# Generate reports
sma_report = generate_report(sma_results)
rsi_report = generate_report(rsi_results)

# Display results
print('SMA Crossover Strategy Report:')
print(sma_report)

print('\nRSI Strategy Report:')
print(rsi_report)
```
